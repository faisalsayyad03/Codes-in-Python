```python
# 4. Convert a string number into an integer and float.

a = 20 

int_num = int(a)
float_num = float(a)

print(int_num)
print(float_num)
```

```python
# 5. Convert an integer into binary, octal, and hexadecimal.

a = 35 

bin_num = bin(a)
oct_num = oct(a)
hex_num = hex(a)

print(bin_num)
print(oct_num)
print(hex_num)
```
