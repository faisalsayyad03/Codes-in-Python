# File Handling

```python
# File handing means reading, writing and managing files using python

# 1. Text File(.txt)
# 2. Binary Files (.bin, .jpg, .mp3)
# 3. CSV File (.csv)
# 4. Json Files (.json)
# 5. Streaming (line by line)
# 6. Full File (all at once)



# "r" → READ only
# (file must exist)
# "w" → WRITE only (creates new / overwrites)
# "a" → APPEND
# (adds to end of file)
# "r+" → READ + WRITE (file must exist)
# "w+" → WRITE + READ (creates new / overwrites)
# =
# → READ binary
# "wb" → WRITE binary
# "ab" → APPEND binary
```
