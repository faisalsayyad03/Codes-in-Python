```python
class BankError(Exception):
    def __init__(self, message):
        super().__init__(message)


class BankAccount:
    
    def __init__(self, account, balance):
        self.__account = str(account)
        self.__balance = balance

    # getters
    def get_account(self):
        return self.__account

    def get_balance(self):
        return self.__balance

    # update account
    def update_account(self, new_account):

        if len(str(new_account)) != 10:
            raise BankError("Please enter 10 digit account number")

        self.__account = str(new_account)

        masked = "*" * 7 + self.__account[-3:]

        print(f"Account Updated Successfully !!")
        print(f"Masked Account Number : {masked}")

    # deposit
    def deposit_money(self, amount):

        if amount <= 0:
            raise BankError("Deposit amount must be greater than zero")

        self.__balance += amount

        print(f"{amount} deposited successfully")
        print(f"Current Balance : {self.__balance}")

    # withdraw
    def withdraw(self, amount):

        if amount <= 0:
            raise BankError("Withdraw amount must be greater than zero")

        if amount > self.__balance:
            raise BankError("Insufficient Balance")

        self.__balance -= amount

        print(f"{amount} withdrawn successfully")
        print(f"Remaining Balance : {self.__balance}")



# object
try:

    b1 = BankAccount(1234567890, 5000)

    b1.deposit_money(2000)
    
    
    b1.withdraw(1000)

    b1.update_account(9076054310)

except BankError as e:
    print(e)
```
